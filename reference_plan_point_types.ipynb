```python
import requests
import pandas as pd

api_url = "https://ows.goszakup.gov.kz/v3/refs/ref_point_type"
token = "73c7ad9a9cbc2611675ec420a44abb12b"
headers = {
    "Authorization": f"Bearer {token}",
    "Content-Type": "application/json"
}
file = "plan_point_type_data.csv"

def retrieve_plan_point_types():
    """Получает данные справочника типов пунктов плана через API и сохраняет в CSV."""
    point_type_list = []

    try:
        params = {"limit": 500}
        response = requests.get(api_url, headers=headers, params=params, timeout=10)
        response.raise_for_status()
        data = response.json()

        items = data.get("items", [])
        if not items:
            print("Нет данных для загрузки.")
        else:
            for item in items:
                point_type_list.append({
                    "id": item.get("id"),
                    "name_kz": item.get("name_kz"),
                    "name_ru": item.get("name_ru")
                })
            print(f"Загружено записей: {len(point_type_list)}")

    except requests.exceptions.RequestException as e:
        print(f"Ошибка при запросе: {e}")

    export_to_csv(point_type_list)

def export_to_csv(dataset):
    """Сохраняет данные в CSV."""
    if not dataset:
        print("Нет данных для сохранения.")
        return
    df = pd.DataFrame(dataset)
    df.to_csv(file, index=False, encoding="utf-8-sig", sep="|", quotechar="'", escapechar="\\")
    print(f"Данные успешно сохранены в файл: {file} ({len(dataset)} записей)")

retrieve_plan_point_types()
```
