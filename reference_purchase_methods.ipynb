```python
import requests
import pandas as pd

api_url = "https://ows.goszakup.gov.kz/v3/refs/ref_trade_methods"
token = "73c7ad9a9cbc2611675ec420a44abb12b"
headers = {
    "Authorization": f"Bearer {token}",
    "Content-Type": "application/json"
}
file = "trade_methods_data.csv"

def retrieve_trade_methods():
    """Получает данные справочника способов закупки через API и сохраняет в CSV."""
    trade_method_list = []

    try:
        params = {"limit": 500}
        response = requests.get(api_url, headers=headers, params=params, timeout=10)
        response.raise_for_status()
        data = response.json()

        items = data.get("items", [])
        if not items:
            print("Нет данных для загрузки.")
        else:
            for item in items:
                trade_method_list.append({
                    "id": item.get("id"),
                    "symbol_code": item.get("symbol_code"),
                    "name_kz": item.get("name_kz"),
                    "name_ru": item.get("name_ru")
                })
            print(f"Загружено записей: {len(trade_method_list)}")

    except requests.exceptions.RequestException as e:
        print(f"Ошибка при запросе: {e}")

    export_to_csv(trade_method_list)

def export_to_csv(dataset):
    """Сохраняет данные в CSV."""
    if not dataset:
        print("Нет данных для сохранения.")
        return
    df = pd.DataFrame(dataset)
    df.to_csv(file, index=False, encoding="utf-8-sig", sep="|", quotechar="'", escapechar="\\")
    print(f"Данные успешно сохранены в файл: {file} ({len(dataset)} записей)")

retrieve_trade_methods()
```
